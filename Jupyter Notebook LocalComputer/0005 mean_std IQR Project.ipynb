```python
# IQR : Interquantile Range
# Q1 :  25th% Value 
# Q3 : 73th% Value
# IQR = Q3 - Q1
#min_value = Q1 - 1.5*IQR
#max_value = Q3 + 1.5*IQR
```

```python
Q1 = df.Sales.quantile(0.25)
Q3 = df.Sales.quantile(0.75) 

Q1  , Q3
```

```python
# IQR  = Q3 - Q1
IQR  = Q3 - Q1
IQR
```

```python
#min_value = Q1 - 1.5*IQR
#max_value = Q3 + 1.5*IQR
min_value = Q1 - 1.5*IQR
max_value = Q3 + 1.5*IQR
min_value , max_value
```

```python
# ye may vala data hata dega (Yaha ek se jyda row hoti to vo bhi hata dega is value 111729.75 se compare karke )
newdf = df[df['Sales']<max_value]
```

```python
#Z_score
# Z = (X-miu)/Sigma
#z =  data point
# miu  = mean
# Sigma = std
```

```python
df['z_score'] = (df['Sales'] - df['Sales'].mean())/df['Sales'].std()
```

```python
df
```

```python
newdf = df[df['z_score']<3]
```
