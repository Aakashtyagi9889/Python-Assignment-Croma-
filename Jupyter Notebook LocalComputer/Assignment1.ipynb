```python
#Q16Create a copy of dataframe.
newdf = df.copy()
```

```python
#Q17Fill missing Salary using average salary.

newdf.Salary = newdf.Salary.fillna(newdf.Salary.mean())
```

```python
#Q18 Fill missing Age using average age.
newdf.Age = newdf.Age.fillna(round(newdf.Age.mean(),2))
```

```python
#Q19 Fill missing Experience using average experience.
newdf.Experience = newdf.Experience.fillna(round(newdf.Experience.mean(),2))
# round(value like 2 and others) ==> it is used to round the values after point values 2.98232783==>2.98
newdf.Experience =  newdf.Experience.round(2)
```

```python
newdf.Experience
```

```python
#Q20Replace missing Name with "Unknown".
newdf.Name = newdf.Name.fillna('Unknown')
```

```python
#Q21 Check whether any null value still exists.
newdf.isnull().sum()
```

```python
#Q22Remove rows having null values.
df = df.dropna()
```
