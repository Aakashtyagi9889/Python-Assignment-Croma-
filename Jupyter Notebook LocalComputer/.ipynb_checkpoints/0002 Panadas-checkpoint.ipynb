```python
data = {
    'Empid': [101,102,103,104,105,106,107,108,109,110],
    'Empname': ['Ram','Shyam','Mohan' , np.nan,'Anu','Riya','Simran','Nameera' ,'Mohit','Bhavna '],
    'Empsalary':[42121,43232,np.nan,43532,83637,37212,np.nan,73536,21341,42313]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#replace
df.fillna("Not Found!")
```

```python
#It works because it does not reflect on original data if you want you can do df = df.fillna("Not Found!")
df.Empsalary.sum()
```

```python
df.Empsalary.fillna('No data')
```

```python
df.Empsalary.fillna(12000)
```

```python
df.Empsalary.fillna(df.Empsalary.mean())
```

```python
newdf = df.copy()
```

```python
newdf.Empsalary = newdf.Empsalary.fillna(round(newdf.Empsalary.mean(),2))
```
