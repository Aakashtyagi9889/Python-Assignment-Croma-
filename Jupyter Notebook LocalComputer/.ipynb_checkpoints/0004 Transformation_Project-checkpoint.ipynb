```python
def convertVotes(val):
    val = str(val)
    if val[-1]=='K':
        return int(float(val[0:len(val)-1])*1000)
    elif val[-1]=='M':
        return int(float(val[0:len(val)-1])*1000000)
    else:
        return int(val)
```

```python
# 24K ,  23.4K , 2M , 2.5M , 344
convertVotes(232)
```

```python
df.Votes = df.Votes.map(convertVotes)
```

```python
def convertDuration(val):
    if len(val.split())==2:
        val = val.split()
        return int(val[0][0:len(val[0])-1])*60 + int(val[1][0:len(val[1])-1])
    elif val[-1] =='h':
        return int(val[0:len(val)-1])*60
    elif val[-1] == 'm':
        return int(val[0:len(val)-1])
    else:
        return convertDuration(df.Duration.mode()[0]) # replacing with mode value {df.Duration.mode()} ==> 1h 30m ==> 90 
```

```python
convertDuration('3213413')
```

```python
df.Duration.mode()
```

```python
df.Duration  = df.Duration.map(convertDuration)
```
