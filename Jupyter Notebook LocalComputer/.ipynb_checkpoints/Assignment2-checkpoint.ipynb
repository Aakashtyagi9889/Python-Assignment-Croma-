```python
#Create DataFrame.
import pandas as pd
import numpy as np

employee = {
'EmpID':[101,102,103,104,105,106,107,108,109,110,111,112],
'Name':['Ram','Shyam',np.nan,'Rohan','Anu','Riya','Simran',
'Neha','Mohit','Bhavna','Amit',np.nan],

'Department':['IT','HR','IT','Finance','Sales',
'HR','IT','Finance','Sales','IT','HR','IT'],

'Salary':[45000,52000,np.nan,65000,
38000,47000,np.nan,72000,41000,58000,49000,np.nan],

'Age':[24,28,np.nan,35,29,
31,27,np.nan,30,36,26,np.nan],

'Experience':[2,5,7,np.nan,
4,6,3,12,np.nan,9,2,5]
}

df = pd.DataFrame(employee)
```

```python
# 2 . Check dataset information.
df.info()
```

```python
#Check null values.
df.isnull()
```

```python
#Count null values
df.isnull().sum()
df.isnull().sum()
```

```python
#Create copy.
newdf = df
```

```python
#Fill Name using "Unknown".
newdf.Name = newdf.Name.fillna('Unknown')
```

```python
newdf
```

```python
#Fill Salary using mean.
newdf.Salary = newdf.Salary.fillna(df.Salary.mean())
```

```python
#Fill Age using mean.
df.Age = df.Age.fillna(round(df.Age.mean(),2))
```

```python
df.isnull().sum()
```

```python
#Fill Experience using mean.
df.Experience = df.Experience.fillna(df.Experience.mean())
```

```python
#Verify no null values remain.
df.isnull().sum()
```

```python
#Total employees
newdf.shape[0]
```

```python
#Average salary
newdf.Salary.mean()
```

```python
#Highest salary
newdf.Salary.max()
```

```python
#Lowest salary
df['Salary'].min()
```

```python
#Average age
newdf.Age.mean()
```

```python
#Average experience
newdf.Experience.mean()
```

```python
#Total salary
newdf.Salary.sum()
```

```python
#Employees earning more than average salary
df[df.Salary>df.Salary.mean()]
```
