```python
df = pd.read_excel(r"C:\Users\DELL\Downloads\Financial_Sample.xlsx")
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype=='str':
            print(col, '\t' , df[col].dtype, '\t\t' , df[col].isnull().sum() , '\tReplaced By : ' , df[col].mode()[0])
            df[col] =  df[col].fillna(df[col].mode()[0])
        else:
            print(col, '\t' , df[col].dtype,'\t', df[col].isnull().sum(), '\tReplaced By : ',round(df[col].mean(),2))
            df[col] = df[col].fillna(round(df[col].mean(),2))
            
   
```
