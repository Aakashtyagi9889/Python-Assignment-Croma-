<h1>MatplotLib</h1>
Matplotlib is a low level ploting library

```python
pbp = df.groupby('Product').agg({'Profit':'sum'}).reset_index()
```

```python
import matplotlib.pyplot as plt
```

```python
pbp
```

```python
matplotlib.__version__
```

```python
plt.bar(pbp['Product'],  pbp['Profit'])
```

```python
plt.figure(figsize = (8,3))
plt.bar(pbp['Product'] , pbp['Profit'])
```

```python
plt.figure(figsize= (8,3))
plt.bar(pbs['Segment'] , pbs.Profit , color = 'pink')
```

```python
plt.figure(figsize = (8 , 3))
plt.bar(pbs.Segment , pbs.Profit ,  color = ['red', 'blue' , 'pink' , 'green', 'yellow'])
```

```python
colors = list(pbs['Profit'].map(lambda val : 'green' if val>0 else 'red'))
```

```python
plt.figure(figsize = (8, 3))
plt.bar(pbs['Segment'] , pbs['Profit'] ,  color = colors)
plt.show()
```

```python
#LineChart
plt.figure(figsize= (8,3))
plt.plot(pbp['Product'] , pbp.Profit)
plt.show()
```

```python
import numpy as np
```

```python
x = np.arange(1, 31 ,1)
y = np.random.randint(100 , 1000 ,30)
```

```python
x , y
```

```python
plt.figure(figsize = (8 ,3))
plt.plot(x , y , 'o')
plt.show()
```

```python
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' )
plt.show()
```

```python
# mfc MArker Face Color 
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red')
plt.show()
```

```python
# ms = Marker Size
# mfc MArker Face Color 
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red' , ms = 8)
plt.show()
```

```python
#mec =  marker edges color
# mfc MArker Face Color 
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red' ,  mec = 'green')
plt.show()
```

```python
# line color change
# mfc MArker Face Color 
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red' ,  mec = 'green' ,  color = 'yellow')
plt.show()
```

```python
#linestyle
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red' ,  mec = 'green' ,  color = 'yellow' , linestyle= 'dashed')
plt.show()
```

```python
#linestyle  = ls
plt.figure(figsize = (8 ,3))
plt.plot(x , y , marker = 'o' , mfc = 'red' ,  mec = 'green' ,  color = 'green' , ls= '-.')
plt.show()
```

```python
#loc= location
plt.figure(figsize = (8, 3))
plt.bar(pbs['Segment'] , pbs['Profit'] ,  color = colors)
plt.xlabel("Name of Segment")
plt.ylabel('Profit in INR (Crore)')
plt.title('Profit of Segment' , loc = 'Right')
plt.show()
```
