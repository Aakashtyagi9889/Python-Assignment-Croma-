```python
# this function show batsman makes highest score at which team 
def run_scored(batsman_name):
    vk = delivery[delivery['batsman']==batsman_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(3).index[0]
```

```python
run_scored('V Kohli')
```

```python
#isin()
#find the most destructive death over batsman in the history of ipl 
#Strike Rate = (Number of Runs / Number of Balls)/100
#min  batsman 200 balls in over 16-20
```

```python
death_over = delivery[delivery['over']>15]
```

```python
all_batsman = death_over.groupby('batsman')['batsman_runs'].count()
x = all_batsman>200
batsman_list = list(all_batsman[x].index)
```

```python
# Now we want to calculate strike rate
# Runs scored  by all these 43 batsman 
# balls played by all these 43 batsman
final = delivery[delivery['batsman'].isin(batsman_list)]
```

```python
runs =final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls =final.groupby('batsman')['batsman_runs'].count()
```

```python
sr = (runs / balls)*100
```

```python
sr
```

```python
sr.sort_values(ascending = False).head(1)
```
