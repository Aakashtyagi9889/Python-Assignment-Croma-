```python
delivery  =  pd.read_csv('deliveries.csv')
```

```python
delivery.head(5)
```

```python
#Find Only those top 5 batsman who has Highest Score 
batsman = delivery.groupby('batsman')
```

```python
batsman['batsman_runs'].sum().sort_values( ascending = False).head(5)
```

```python
#Find Only those top 5 batsman who has Highest 4 boundaries
mask =  delivery['batsman_runs']==4
new_delivery = delivery[mask]
```

```python
# Find All Four Boundaries 
new_delivery.shape[0]
```

```python
new_delivery.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head()
```

```python
vk = delivery[delivery['batsman']=='V Kohli']
```

```python
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(3)
```

```python
# this function show batsman makes highest score at which team 
def run_scored(batsman_name):
    vk = delivery[delivery['batsman']==batsman_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(3).index[0]
```

```python
run_scored('V Kohli')
```
